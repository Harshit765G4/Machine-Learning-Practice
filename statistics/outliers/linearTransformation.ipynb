```python
import numpy as np
```

```python
# Input: Pack Vertices into One Matrix

R = np.array([
    [0, -1],
    [1, 0]
])
```

```python
P = np.array([
    [1, 3, 2],
    [1, 1, 2]
])
```

```python
# Calculation: One Operation Transforms all Points

# The @ symbol is pythons operator for matrix multiplication

P_transformed = R @ P
```

```python
print(P)
```

```python
print(P_transformed)
```
