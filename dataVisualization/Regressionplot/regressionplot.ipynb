```python
# A regression plot shows the relationship between two numerical variables and draws a line that represents their overall trend.

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
```

```python
sns.lmplot(x='total_bill', y='tip',data=tips,hue='sex')
```

```python
sns.lmplot(x='total_bill', y='tip',data=tips,hue='sex',palette='coolwarm')
```

```python
sns.lmplot(x='total_bill', y='tip',data=tips,hue='sex',palette='coolwarm',markers=['o','v'],scatter_kws={'s':15})
```

```python
sns.lmplot(x='total_bill', y='tip',data=tips,hue='sex',palette='coolwarm',markers=['o','v'],scatter_kws={'s':150})
```
