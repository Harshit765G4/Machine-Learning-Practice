```python
# Matplotlib Basics

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
```

```python
y = x ** 3
y
```

```python
plt.plot(x,y)
plt.title('Plot Of X and Y')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

```python
# SubPlots in Matplotlib
plt.subplot(2,2,1)
plt.plot(x,y)

plt.subplot(2,2,4)
plt.plot(y,x)

plt.subplot(2,2,3)
plt.plot(y*2,x**2)

plt.subplot(2,2,2)
plt.plot(x**2,y**3)

```

```python
# Object oriented Way

# - Matplotlib allows the aspect ratio, DPI and figure size to be specified when the figure object is created. you can use the figsize and spi keyword arguments.

# - figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inches (pexels per inch).

fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1]) # [left,bottom,width,height]
axis1.plot(x,y)
axis2 = fig.add_axes([0.2,0.6,0.5,0.4])
axis2.plot(x,y)
```

```python
fig = plt.figure(figsize=(12,8), dpi=100)
axis1 = fig.add_axes([0.1,0.1,0.8,0.8]) # [left,bottom,width,height]
axis1.plot(x,y)
```

```python
# Types of Plots

# 1. Scatter Plot 

plt.scatter(x,y)
```

```python
# 2. Histogram

from random import sample
data = sample(range(1,1000),100)
plt.hist(data)
```

```python
# 3. Box Plot 
plt.boxplot(x)
```

```python
# Saving Plots

fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax .plot(x,y)

plt.savefig('BasicPlotOfXandY.png')
```
