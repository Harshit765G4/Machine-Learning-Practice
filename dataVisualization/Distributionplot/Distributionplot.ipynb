```python
# Distribution Plot

# We use distribution plots to visualize the distribution of quantitative data.
# these Plots are:- 
# 1. distplot/histplot
# 2. joinplot
# 3. pairplot
# 4. rugplot



import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the Data

df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

```python
# Distplot / Histplot - A histogram is a plot used to understand the distribution of numerical data.

sns.displot(df['total_bill'])
```

```python
# bins is the value that helps us to plot, it is a range/group of numerical values.

sns.displot(df['total_bill'],bins=20)
```

```python
sns.displot(df['total_bill'],bins=100)
```

```python
# Kernel Density Estimate - KDE creates a smooth curve that shows where your numerical data is concentrated.

sns.displot(df['total_bill'],bins=20, kde= True)
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=20, kde= True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20, kde= True)
```

```python
# joinplot - A joint plot shows the relationship between two numerical variables.

# kind - scatter

sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'scatter')
```

```python
# kind - hex

sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'hex')
```

```python
# kind - kde

sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'kde')
```

```python
# kind - hist

sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'hist')
```

```python
# kind - reg

sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'reg')
```

```python
# kind - resid

sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'resid')
```

```python
# Pairplot - A pair plot shows the relationships and distributions of multiple numerical columns in a DataFrame.

sns.pairplot(df)
```

```python
sns.pairplot(df,hue='sex')
```

```python
sns.pairplot(df,hue='sex', palette='rainbow')
```

```python
sns.pairplot(df,hue='size', palette='rainbow')
```

```python
# Rugplot - A rug plot shows the location of every individual data point on a number line.

sns.rugplot(df)
```

```python
sns.rugplot(df['total_bill'])
```
