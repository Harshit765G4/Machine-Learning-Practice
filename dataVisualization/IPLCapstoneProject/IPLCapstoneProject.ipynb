```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings 
warnings.filterwarnings('ignore')

df = pd.read_csv('IPL.csv')

df.head(3)
```

```python
# Check the size of rows and columns of the dataset

print(f'your rows are {df.shape[0]} and your columns are {df.shape[1]}')
```

```python
df.head(1)
```

```python
# 1. Which team won the most matches?

match_wins = df['match_winner'].value_counts()

sns.barplot(y=match_wins.index, x=match_wins.values, palette='rainbow')

plt.title("Most Matches Win by Team")
```

```python
# Toss Decision Trend

df.head(2)
```

```python
sns.countplot(x=df['toss_decision'], palette='rainbow')
plt.title("Toss Decision Trend")
```

```python
# 3. Toss Winner Vs Match Winner

count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count * 100)/df.shape[0]
print(percentage.round(2))
```

```python
# # Key Player Performences

# 1. Most "Player of the Match" Awards

df.head(1)
```

```python
count = df['player_of_the_match'].value_counts().head(10)
sns.barplot(x= count.values, y= count.index, palette="mako")
plt.title("Top 10 Man of the Match")
```

```python
# 2 Top Scorres

df.head(2)
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind='barh')
```

```python
# 10 Best Bowling Figures

df.head(3)
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x: x.split('--')[0])
```

```python
df['highest_wickets'] = df['highest_wickets'].astype(int)
```

```python
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
```

```python

top_bowlers
```

```python
top_bowlers.plot(kind='barh')
```

```python
# Venue Analysis

# Most Matches Played by Venue

venue_count = df['venue'].value_counts()
venue_count
```

```python
sns.barplot(x= venue_count.values, y= venue_count.index, palette='rainbow')
```
