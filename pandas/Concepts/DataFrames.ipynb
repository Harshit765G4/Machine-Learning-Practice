```python
# DataFrames: Combining Multiple Series together it Becomes a DataFrame

# Creating a DataFrame

import numpy as np
import pandas as pd
```

```python
data = {
    'Name': ['John','Anna','Peter','Linda'],
    'Age': [28,34,29,42],
    'City': ['New York','Paris','Berlin','London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
df
```

```python
data_list = [
    ['John',28,'New York',65000],
    ['Anna',34,'Paris',70000],
    ['Peter', 29,'Berlin',62000],
    ['Linda', 42, 'London', 85000]
]
labels = ['Name', 'age', 'City', 'Salary']
df2 = pd.DataFrame(data_list, columns = labels)
df2
```

```python
# Selection and Indexing of columns

df2
```

```python
df2['Name']
```

```python
df2['City']
```

```python
df2[['Name','City']]
```

```python
# Creating a New Columns

df2['Designation'] = ['Software Engineer','Data Analyst', 'Doctor', 'Product Engineer']
df2
```

```python
# Removing a Column

df2.drop('Designation', axis = 1,inplace = True)
```
