```python
# Pandas Series 
# - A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

# Use Cases of Pandas
# - Data Cleaning
# - Data Analysis
# - Data Transformation
# - Data Visualization(Basic Level)
# - Data Aggregation
# - File Handling
# - Data Filtering & Selection
# - Time Series Analysis

import numpy as np
import pandas as pd
```

```python
# Creating Series - There are multiple ways to create a pandas Series:

labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {1:10,2:20,3:30}
```

```python
pd.Series(my_list)
```

```python
# Here in the above output 
# 0    10
# 1    20
# 2    30
# dtype: int64

# 0 1 2  are labels or index points
# 10 20 30 are data points
```

```python
pd.Series(my_list, index = labels)
```

```python
pd.Series(arr, index = labels)
```

```python
pd.Series(d)
```
